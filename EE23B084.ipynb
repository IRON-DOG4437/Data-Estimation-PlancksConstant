```python
import csv
import numpy as np
L = []
B = []
with open("d1.csv", "r",  newline='') as f:
    data_set = csv.reader(f)
    for row in data_set:
        row = list(map(float, row))
        L.append(row[0])
        B.append(row[1])
```

```python
import matplotlib.pyplot as plt

plt.plot(L, B)
```

```python
print(len(L))
```

```python
###Using brute force approach
```

```python
def equation(L, h, c, k,T):

    try:
        coeff = 2 * h * (c ** 2) / (L**5)
        fxn = 1 / (np.exp(h*c/(k * T * L)) - 1)

    except ZeroDivisionError:
        print("#")
        coeff = 1
        fxn = 1
    # print(L, coeff, fxn)
    return coeff * fxn

def partial_equation(L,T):
    h_fixed = 6.6e-34
    c_fixed = 3e+8
    k_fixed = 1.3e-23
    return equation(L,h_fixed,c_fixed,k_fixed,T)
```

```python
def loss_function(Y_predicted, Y_given):
    square = np.vectorize(lambda x: x**2)
    return np.sum(square(Y_predicted - Y_given)) / len(Y_given)
```

```python
#Manually assumed approx Target values 
h = 6.6e-34
c = 3e8
k = 1.3e-23
T = 4250
Y = []
#X = np.array([i * 1e-7 / 2 for i in range(1, 101)])
X = np.array(L)
#partial equation
eq_vec = np.vectorize(partial_equation)
Y = eq_vec(X, T)*1e-3

#Full equation
eq_vec_2 = np.vectorize(equation)
Y_2 = eq_vec_2(X, h,c,k,T)*1e-3
```

```python
print(loss_function(Y, B))
```

```python
from scipy.optimize import curve_fit
#Partial application
popt, pcov = curve_fit(partial_equation, L, B, p0 = [1e+3])

#Full application
popt_2, pcov_2 = curve_fit(equation, L, B, p0 = [1e-34, 1e8, 1e-23, 1e3])
```

```python
print(popt)
```

```python
#Partial Application
Y_opt = eq_vec(X, popt[0])

#Full application
Y_opt_2 = eq_vec_2(X, popt_2[0], popt_2[1], popt_2[2], popt_2[3])


plt.plot(L, B)
plt.plot(X, Y_opt)
plt.show()
plt.plot(L, B)
plt.plot(X, Y_opt_2)
plt.show()
```
